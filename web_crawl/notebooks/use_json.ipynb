```python
import json
```

JSON, or JavaScript Object Notation, is a widely-used text-based format for data interchange. Its syntax resembles Python dictionaries but with some differences, such as using only double quotes for strings and lowercase for Boolean values.

JSON’s straightforward syntax allows both humans and computers to read and write JSON data effortlessly.

```python
# very basic JSON example
{
  "greeting": "Hello, world!"
}
```

# Examining JSON syntax

```python
{
  "name": "Frieda",
  "isDog": True,
  "hobbies": ["eating", "sleeping", "barking"],
  "age": 8,
  "address": {
    "work": None,
    "home": ["Berlin", "Germany"]
  },
  "friends": [
    {
      "name": "Philipp",
      "hobbies": ["eating", "sleeping", "reading"]
    },
    {
      "name": "Mitch",
      "hobbies": ["running", "snacking"]
    }
  ]
}
```

The JSON standard doesn’t allow any comments, trailing commas, or single quotes for strings.

# Writing JSON with Python

```python
food_ratings = {"organic dog food": 2, "human food": 10}
json.dumps(food_ratings) # converts dict into json string
```

```python
str(food_ratings)
```

```python
numbers_present = {1: True, 2: True, 3: False}
json.dumps(numbers_present)
```

```python
# takes care of all python to JSON conversions

dog_id = 1
dog_name = "Frieda"
dog_registry = {dog_id: {"name": dog_name}}
json.dumps(dog_registry)
```

Note that different Python data types like lists and tuples serialize to the same JSON array data type. This can cause problems when you convert JSON data back to Python, as the data type may not be the same as before.

You can’t use dictionaries, lists, or tuples as JSON keys.

By providing the `skipkeys` argument, you can prevent getting a TypeError when creating JSON data with unsupported Python keys. When you set skipkeys in json.dumps() to True, then Python skips the keys that are not supported and would otherwise raise a TypeError. The result is a JSON-formatted string that only contains a subset of the input dictionary.



```python
# when using json.dumps(), you can use additional arguments to control the look of the resulting JSON-formatted string

toy_conditions = {"chew bone": 7, "ball": 3, "sock": -1}
json.dumps(toy_conditions, sort_keys=True)
```

# Writing a JSON file with Python

```python
# To write Python data into an external JSON file, you use json.dump()

dog_data = {
  "name": "Frieda",
  "is_dog": True,
  "hobbies": ["eating", "sleeping", "barking",],
  "age": 8,
  "address": {
    "work": None,
    "home": ("Berlin", "Germany",),
  },
  "friends": [
    {
      "name": "Philipp",
      "hobbies": ["eating", "sleeping", "reading",],
    },
    {
      "name": "Mitch",
      "hobbies": ["running", "snacking",],
    },
  ],
}

# recall: context manager
with open("hello_frieda.json", mode="w", encoding="utf-8") as write_file:
    json.dump(dog_data, write_file)
```

# Reading JSON with Python

json.loads(): To deserialize a string, bytes, or byte array instances

json.load(): To deserialize a text file or a binary file

```python
# converting JSON objects to python dictionary

dog_registry = {1: {"name": "Frieda"}}
dog_json = json.dumps(dog_registry)
dog_json

new_dog_registry = json.loads(dog_json) # python dict
new_dog_registry
```

```python
dog_registry # json object
```

```python
# dictionary =/= json object

new_dog_registry == dog_registry
```
