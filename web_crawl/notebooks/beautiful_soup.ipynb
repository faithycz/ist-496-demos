# Step 3: Parse HTML code with Beautiful Soup

```python
URL = "https://realpython.github.io/fake-jobs/"
page = requests.get(URL)

# creating a beautiful soup object: page.content is the HTML content, html.parser makes sure you're using an appropriate parser for html content
soup = BeautifulSoup(page.content, "html.parser")
soup
```

```python
# finding specific HTMl elements by ID

results = soup.find(id="ResultsContainer")
results.prettify()
```

```python
# finding HTML elements by class name

job_cards = results.find_all("div", class_="card-content")
```

```python
for job_card in job_cards:
    print(job_card, end="\n" * 2)
```

# Extracting text from HTML elements

```python
for job_card in job_cards:
    title_element = job_card.find("h2", class_="title")
    company_element = job_card.find("h3", class_="company")
    location_element = job_card.find("p", class_="location")
    print(title_element.text.strip())
    print(company_element.text.strip())
    print(location_element.text.strip())
    print()
```

# Find elements by class name and text content

```python
# Not all of the job listings are developer jobs. Instead of printing out all the jobs listed on the website, you’ll first filter them using keywords.
# to filter for specific jobs/keywords, use the string argument

# looks for the EXACT string, uh oh! may not work for other variations!
python_jobs = results.find_all("h2", string="Python")
python_jobs
```

```python
# you can sometimes pass functions as arguments to Beautiful Soup methods

# passing an anonymous function to the string argument...
# lambda function looks at the text of each <h2> element, converts it to lowercase, and checks whether the substring "python" is found anywhere
python_jobs = results.find_all(
    "h2", string=lambda text: "python" in text.lower()
)

# it works now!
len(python_jobs)
```

# Identifying error conditions

```python
# when you try to print the filtered python jobs now... you get an error
# you only filtered for h2 elements that included the word "python"
# when you try to find more info, python_jobs only includes job title text
python_jobs[0]
```

# Accessing parent elements

```python
# one way to get access to all the information for a job is to step up in the hierarchy of the DOM starting from the <h2> elements that you identified.
# find the <h2> element that contains the job title and its closest parent element that contains the information you’re interested in...

python_jobs = results.find_all(
    "h2", string=lambda text: "python" in text.lower()
 )

# list comprehension, looking at third-level parent
python_job_cards = [
    h2_element.parent.parent.parent for h2_element in python_jobs
]
```

```python
# iterating over parent elements instead

for job_card in python_job_cards:
    title_element = job_card.find("h2", class_="title")
    company_element = job_card.find("h3", class_="company")
    location_element = job_card.find("p", class_="location")
    print(title_element.text.strip())
    print(company_element.text.strip())
    print(location_element.text.strip())
    print()
```

# Extracting attributes from HTML elements

```python
# fetching job links... but text
for job_card in python_job_cards:
    links = job_card.find_all("a")
    for link in links:
        print(link.text.strip())
```

```python
# proper way to fetch links (using a attribute)

for job_card in python_job_cards:
    links = job_card.find_all("a")
    for link in links:
        link_url = link["href"]
        print(f"Apply here: {link_url}\n")
```

```python
# to look for the second link element...

for job_card in python_job_cards:
    link_url = job_card.find_all("a")[1]["href"]
    print(f"Apply here: {link_url}\n")
```

```python
# sample scaper.py script
# a Python web scraping script that extracts the job title, company, location, and application link from the scraped website

URL = "https://realpython.github.io/fake-jobs/"
page = requests.get(URL)

soup = BeautifulSoup(page.content, "html.parser")
results = soup.find(id="ResultsContainer")

python_jobs = results.find_all(
    "h2", string=lambda text: "python" in text.lower()
)

python_job_cards = [
    h2_element.parent.parent.parent for h2_element in python_jobs
]

for job_card in python_job_cards:
    title_element = job_card.find("h2", class_="title")
    company_element = job_card.find("h3", class_="company")
    location_element = job_card.find("p", class_="location")
    print(title_element.text.strip())
    print(company_element.text.strip())
    print(location_element.text.strip())
    link_url = job_card.find_all("a")[1]["href"]
    print(f"Apply here: {link_url}\n")
```
