# List and dict comprehensions

```python
fruits = ["Apples", "Bananas", "Pears"]

book1 = {"title": "Great Expectations", "author": "Charles Dickens"}
book2 = {"title": "Bleak House", "author": "Charles Dickens"}
book3 = {"title": "A Book By No Author"}
book4 = {"title": "Moby Dick", "author": "Herman Melville"}

books = [book1, book2, book3, book4]
```

```python
for fruit in fruits:
    print(fruit)
```

```python
fruits_shouted = []
for fruit in fruits:
    fruits_shouted.append(fruit.upper())

fruits_shouted
```

```python
# list comprehension - a concise way to create a new list by applying an expression to each item in an existing iterable (e.g. a list)

fruits_shouted2 = [fruit.upper() for fruit in fruits]
fruits_shouted2
```

```python
# dictionary comprehension - a concise way to create a new dictionary from an existing iterable

fruit_mapping = {fruit: fruit.upper() for fruit in fruits}
fruit_mapping
```

```python
# use if statement to filter results

fruits_with_longer_names_shouted = [fruit.upper() for fruit in fruits if len(fruit)>5]
fruits_with_longer_names_shouted
```

```python
fruit_mapping_unless_starts_with_a = {fruit: fruit.upper() for fruit in fruits if not fruit.startswith('A')}
fruit_mapping_unless_starts_with_a
```

```python
# another comprehension ("go through every book in books, and put the titles in a new list")

[book['title'] for book in books]
```

```python
# fails because one of the books doesn't have an author

[book['author'] for book in books]
```

```python
# get() returns none if the value doesn't exist, so no errors

[book.get('author') for book in books]
```

```python
# filters out the "None" value

[book.get('author') for book in books if book.get('author')]
```

```python
# converts into set, removes duplicates

set([book.get('author') for book in books if book.get('author')])
```

```python
# set comprehension

{book.get('author') for book in books if book.get('author')}
```

# Generators
Generator - a special type of function that returns an iterator object, allowing you to produce a sequence of values over time on-demand instead of computing them all at once. Helps with memory efficiency.

When a function has `yield` instead of `return`, it becomes a generator function.
`yield` pauses the function, saves its state, and returns a value to the caller.

```python
import time

def come_up_with_fruit_names():
    for fruit in fruits:
        time.sleep(1) # thinking of a fruit
        yield fruit
```

```python
for fruit in come_up_with_fruit_names():
    print(fruit)
```

```python
def authors_generator():
    for book in books:
        if book.get("author"):
            yield book.get("author")
```

```python
for author in authors_generator():
    print(author)
```

```python
# using list comprehension

def authors_generator():
    for author in [book.get("author") for book in books if book.get("author")]:
        yield author
```

```python
for author in authors_generator():
    print(author)
```

```python
# shortcut version

def authors_generator():
    yield from [book.get("author") for book in books if book.get("author")]
```

```python
for author in authors_generator():
    print(author)
```

```python
# using set comprehension

def unique_authors_generator():
    yield from {book.get("author") for book in books if book.get("author")}
```

```python
for author in unique_authors_generator():
    print(author)
```

```python
import random
import time

pronouns = ["I", "You", "We", "They"]
verbs = ["eat", "detest", "bathe in", "deny the existence of", "resent", "pontificate about", "juggle", "impersonate", "worship", "misplace", "conspire with", "philosophize about", "tap dance on", "dramatically renounce", "secretly collect"]
adjectives = ["turqoise", "smelly", "arrogant", "festering", "pleasing", "whimsical", "disheveled", "pretentious", "wobbly", "melodramatic", "pompous", "fluorescent", "bewildered", "suspicious", "overripe"]
nouns = ["turnips", "rodents", "eels", "walruses", "kumquats", "monocles", "spreadsheets", "bagpipes", "wombats", "accordions", "mustaches", "calculators", "jellyfish", "thermostats"]

def infinite_random_sentences():
    while True:
        yield from random.choice(pronouns)
        yield " "
        yield from random.choice(verbs)
        yield " "
        yield from random.choice(adjectives)
        yield " "
        yield from random.choice(nouns)
        yield ". "

for letter in infinite_random_sentences():
    print(letter, end="", flush=True)
    time.sleep(0.02)
```

```python
# Write a Book class with a title and author. Include a method has_author()
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def has_author(self):
        return self.author is not None

# Write a BookShelf class with a list of books. Include a generator method unique_authors()
class BookShelf:
    def __init__(self, books):
        self.books = books

    def unique_authors(self):
        authors = set()

        for book in self.books:
            if book.has_author():
                authors.add(book.author)

        for author in authors:
            yield author
```

```python
book = Book("Great Expectations", "Charles Dickens")

print(book.has_author())
```

```python
books2 = [Book(book["title"], book.get("author")) for book in books]

bookshelf = BookShelf(books2)

print([author for author in bookshelf.unique_authors()])
```

# Subclasses, Type Hints, Pydantic

Subclass - a class that inherits attributes and methods from another class, known as the superclass.

Type Hints - an optional syntax in Python that lets you explicitly state the expected data types of variables, function parameters, and return values.

Pydantic - a data validation and parsing library for Python, uses standard Python type hints to enforce data types and structure at runtime. Built around defining schemas as Python classes, ensures the data matches the expected types, and optionally transforms or validates the data automatically.

```python
import pydantic
```

```python
# basic example

class Book:
    title: str
    author: str

    def __init__(self, title: str, author: str):
        self.title = title
        self.author = author
```

```python
# optional hints at the chance an attribute may be "None"

from typing import Optional

class Book:
    title: str
    author: Optional[str] = None  # author may not be known at first
```

```python
# union hints at the chance an attribute may be multiple types

from typing import Union

class Book:
    title: str
    year_published: Union[int, str]  # year_published could be "Unknown" or an int
```

```python
# Subclass - allows you to create a class that inherits attributes and methods from another class. Useful when you want to build on existing functionality.

class Book:
    def __init__(self, title: str, author: str):
        self.title = title
        self.author = author

    def get_description(self) -> str:
        return f"'{self.title}' by {self.author}"

# subclass for eBooks
class EBook(Book):
    def __init__(self, title: str, author: str, file_format: str):
        super().__init__(title, author)  # initialize attributes from Book
        self.file_format = file_format

    def get_description(self) -> str:
        # overriding the parent class method
        return f"'{self.title}' by {self.author} (format: {self.file_format})"
```

```python
# Class methods - methods that belong to the class itself, rather than any instance of the class. Can be used to directly call the class without creating an instance first.
# Commonly used for factory methods, which provide different ways to instantiate objects based on varied input formats.

from typing import List

class Book:
    title: str
    author: str

    def __init__(self, title: str, author: str):
        self.title = title
        self.author = author

    def get_description(self) -> str:
        return f"'{self.title}' by {self.author}"

    # "cls" is the first parameter that represents the class, unlike "self" which represents the instance
    @classmethod
    def create_multiple(cls, titles_and_authors: List[tuple]) -> List['Book']:
        """Class method that returns a list of Book instances from a list of title-author pairs."""
        return [cls(title, author) for title, author in titles_and_authors]
```

```python
books = Book.create_multiple([("1984", "George Orwell"), ("Brave New World", "Aldous Huxley")])
for book in books:
    print(book.get_description())
```

```python
# Basic Pydantic example
# All Pydantic models inherit from pydantic.BaseModel. When you create a subclass of BaseModel, you define a data schema.

from pydantic import BaseModel

class Book(BaseModel):
    title: str
    author: str
    pages: int
    published: bool = True  # default value

book = Book(title="1984", author="George Orwell", pages=328)
print(book)
```

```python
# Pydantic validates types and even tries to coerce values when possible.

book = Book(title="Dune", author="Frank Herbert", pages="412")
print(book.pages)  # int: 412, even though input was a string
```

```python
# if something really doesn't work... throws an error

Book(title="Dune", author="Frank Herbert", pages="not_a_number")
```

```python
# instantiating objects from a dictionary

fields = {
    "title": "The Hobbit",
    "author": "J.R.R. Tolkien",
    "pages": 310
}

book = Book(**fields)
print(book)
```

```python
# converts model into dictionary representation using model_dump()

print(book.model_dump())
```

```python
# convert model to JSON string

print(book.model_dump_json())
```

```python
# loading model from JSON using json library
import json

json_string = '{"title": "Frankenstein", "author": "Mary Shelley", "pages": 280}'
data = json.loads(json_string)
book = Book(**data)
```

```python
# nested models

class Author(BaseModel):
    name: str
    birth_year: int

class Book(BaseModel):
    title: str
    author: Author
    pages: int

a = Author(name="Isaac Asimov", birth_year=1920)
b = Book(title="Foundation", author=a, pages=255)
print(b)
```

```python
# building from nested dictionaries

data = {
    "title": "Neuromancer",
    "author": {
        "name": "William Gibson",
        "birth_year": 1948
    },
    "pages": 271
}

b = Book(**data)
print(b)
```

# Decorators
Decorator - a function that modifies the behavior of another function or method. It allows for the addition of functionality to existing code in a clean and readable manner without altering the original function's structure.

```python
# set_raise_amount modifies the class variable raise_amount, affecting all instances of Employee

class Employee:
    raise_amount = 1.05

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    # decorator
    @classmethod
    def set_raise_amount(cls, amount):
        cls.raise_amount = amount

Employee.set_raise_amount(1.10)
print(Employee.raise_amount)
```

```python
# @staticmethod is a method that doesn't access or modify class or instance state

class MathOperations:
    # decorator
    @staticmethod
    def add(x, y):
        return x + y

result = MathOperations.add(5, 3)
print(result)
```

```python
# custom logging decorator

import functools

def log_decorator(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        print(f"Calling function: {func.__name__}")
        result = func(*args, **kwargs)
        print(f"Function {func.__name__} completed")
        return result
    return wrapper

@log_decorator
def greet(name):
    print(f"Hello, {name}!")
```

```python
greet("Alice")
```

# Docker

Docker - a platform that enables developers to package applications and their dependencies into containers. This ensures consistent performance across different environments.

Container - a lightweight, standalone executable package that includes everything needed to run a piece of software.

Image - a read-only template used to create containers.

Docker Engine - the core software that runs and manages containers.
