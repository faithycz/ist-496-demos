# Python Files

```python
import os
from pathlib import Path
import glob
```

```python
# open a file for reading, writing, or other operations
# read (r)
with open("testfile.txt", "r", encoding="utf-8") as file:
    content = file.read()
    print(content)
```

```python
# write (w)
with open("testfile.txt", "w", encoding="utf-8") as file:
    file.write("Hello!")
```

```python
# append (a)
with open("testfile.txt", "a", encoding="utf-8") as file:
    file.write("\nNew line!")
```

```python
# exclusive creation (x) -- creates new file
with open("newfile.txt", "x", encoding="utf-8") as file:
    file.write("Created!")
```

```python
# context manager (with open) automatically closes the file when done
with open("testfile.txt", "r", encoding="utf-8") as file:
    content = file.read()
    print(content)
```

```python
# reading file line by line instead of loading entire file from memory (for large files)
with open("testfile.txt", "r", encoding="utf-8") as file:
    for line in file:
        print(line.strip())  # Strip removes newline characters
```

# Path Handling

```python
# constructs a file path using os.path
file_path = os.path.join("folder", "subfolder", "example.txt")
print(file_path)
```

```python
# constructs a file path using pathlib library
file_path = Path("C:\\") / "Users" / "faith" / "python-review"
print(file_path)
```

```python
# checks if a file exists
if file_path.exists():
    print(f"{file_path} exists!")
```

# Listing files in a directory

```python
# lists all .txt files in the current directory
txt_files = glob.glob("*.txt")
print(txt_files)
```

```python
# lists all .txt files in current directory and subdirectories
all_txt_files = glob.glob("**/*.txt", recursive=True)
print(all_txt_files)
```
