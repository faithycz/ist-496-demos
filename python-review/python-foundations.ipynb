# Python Functions

```python
def calculate_area(width, height):
    area = width * height
    return area

area1 = calculate_area(5, 10)
print("Area of rectangle 1:", area1)

area2 = calculate_area(3, 4)
print("Area of rectangle 2:", area2)
```

```python
def calc2(width: int, height: int):
    return width * height

print(calc2(3, 5))
# print(calc2("a", "b"))
```

# Python Lists

```python
# Creating a list of fruits
fruits = ["apple", "banana", "cherry"]
```

```python
# Accessing items
print(fruits[0])
print(fruits[1])
```

```python
# Adding an item
fruits.append("orange")
print(fruits)
```

```python
# Removing an item
fruits.remove("banana")
print(fruits)
```

# Python Dictionaries

```python
# Creating a dictionary with some information about a person
person = {
    "name": "Alice",
    "age": 25,
    "city": "New York"
}
```

```python
# Accessing values by key
print(person["name"])
print(person["age"])
```

```python
# Adding a new key-value pair
person["job"] = "Engineer"
print(person)
```

```python
# Changing a value
person["city"] = "San Francisco"
print(person)
```

```python
# Using get() without a default value
print(person.get("name"))
print(person.get("country"))
```

```python
# Using get() with a default value
print(person.get("country", "USA"))
```

```python
list(person.keys())
```

```python
list(person.values())
```

```python
for key, value in person.items():
    print(key, value)
```
