# Python Dictionaries

```python
fruits = ["Apples", "Bananas", "Pears"]

fruit_dict = {index:value for index, value in enumerate(fruits)}
fruit_dict
```

```python
fruit_dict.keys()
```

```python
fruit_dict.values()
```

```python
fruit_dict.items()
```

```python
def create_profile(username, role, status):
    print(f"User: {username} | Role: {role} | Status: {status}")

# The dictionary keys match the function parameters
user_data = {
    "username": "Alice",
    "role": "Admin",
    "status": "Active"
}

# Unpacking the dictionary using **
create_profile(**user_data)
```

```python
def create_profile2(username: str, role: str, status: str) -> None:
    print(f"User: {username} | Role: {role} | Status: {status}")
```

```python
create_profile2(10, 9, 8)
```

```python
def create_profile3(username: str, role: str, status: str) -> str:
    return f"User: {username} | Role: {role} | Status: {status}"
```
