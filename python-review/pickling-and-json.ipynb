# Pickling Python Objects
Pickling is a way to serialize (convert) a Python object into a binary format so that it can be saved to a file and later reloaded. This is especially useful when you want to save complex objects. _It's faster than converting objects to text formats, like JSON._

Pickling Steps:
1. Serialize an object (save it to a file)
2. Deserialize an object (load it back into memory)

```python
import pickle
import json
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# repr() returns a string representation of an object
    def __repr__(self):
        return f"Person(name={self.name}, age={self.age})"
```

```python
# pickling an object

# create an instance of the class
person = Person("John", 20)

# open a file in binary write mode
with open("person.pkl", "wb") as file:
    pickle.dump(person, file)  # serialize and write to file

print("Object pickled successfully.")
```

```python
# unpickling an object (reading the object back)

# open the file in binary read mode
with open("person.pkl", "rb") as file:
    loaded_person = pickle.load(file)  # deserialize and load

print("Loaded object:", loaded_person)
```

# JSON Serialization
An alternative to pickling; only works with basic data types.

```python
data = {
    "name": "John",
    "age": 20,
    "hobbies": [
        "reading",
        "writing"
    ]
}

# save simple dictionary to a JSON file
with open("data.json", "w") as file:
    json.dump(data, file)

print("Data saved as JSON.")
```

```python
# loading back the JSON data

# read from the JSON file
with open("data.json", "r") as file:
    loaded_data = json.load(file)

print("Loaded JSON data:", loaded_data)
```

```python
# convert python object to JSON string
json_string = json.dumps(data)
print("JSON String:", json_string)

# convert JSON string back to python object
python_obj = json.loads(json_string)
print("Python Object:", python_obj)
```

**Note:** Use pickling for saving objects locally (and speed + flexibility), and use JSON for sharing data with other systems or languages.
