```python
import os
from dotenv import load_dotenv, dotenv_values
```

```python
load_dotenv() # this load everything from .env to the OS environment, the above essentially reads from .env, then invokes os.setenv(...)
```

```python
color = os.getenv("COLOR") # now you can read variable from the OS environment
# if you don't want to insert everything from .env into your OS environment
# do the following
# this loads everything into .env into a Python dictionary
config = dotenv_values(".env")
color = config["COLOR"] # now you read from the dictionary
print(color)
```
