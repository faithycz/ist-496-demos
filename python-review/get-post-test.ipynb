```python
import requests
import json
import os # for env variable
```

# GET Request Test

```python
# API endpoint
url = "https://reqres.in/api/collections/products/records?project_id=41054"

# getting env variable
key_var = os.getenv("TEST_API_KEY")

# headers
request_headers = {
    "x-api-key": "free_user_3H6aCg0SRgCQgektNkfOl4NoWl0"
    # "x-api-key": key_var
}

# send GET request
response = requests.get(url, headers=request_headers)

# parse API response
if response.status_code == 200:
    data = response.json()
    json_beautify = json.dumps(data, indent=3)
    print(json_beautify)
else:
    print(f"Error {response.status_code}: {response.text}")
```

# POST Request Test

```python
# API endpoint
url = "https://reqres.in/api/collections/products/records?project_id=41054"

key_var = os.getenv("TEST_API_KEY")

# headers
request_headers = {
    # "x-api-key": "free_user_3H6aCg0SRgCQgektNkfOl4NoWl0"
    "x-api-key": key_var
}

# payload
payload = {
    "data": {
    "name": "Seedless Watermelon",
    "price": 9.99,
    "category": "Fruit",
    "in_stock": True
  }
}

# send POST request
response = requests.post(url, headers=request_headers, json=payload)

# parse API response
if response.status_code == 201:
    data = response.json()
    json_beautify = json.dumps(data, indent=3)
    print(json_beautify)
else:
    print(f"Failed with status: {response.status_code}")
```
