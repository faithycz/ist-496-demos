```python
# Creating a list of fruits
fruits = ["apple", "banana", "cherry"]
```

```python
[print(fruit) for fruit in fruits]
```

```python
# Accessing items
print(fruits[0])
print(fruits[1])
```

```python
# Adding an item
fruits.append("orange")
print(fruits)
```

```python
# Removing an item
fruits.remove("banana")
print(fruits)
```

```python
def print_item(some_list):
    for item in some_list:
        print(item)
```

```python
print_item(fruits)
```

```python
print_item(some_list=fruits)
```

```python
def add(x, y, z):
    return x + y + z
```

```python
add(1, 2, 3)
```

# List Comprehensions

```python
numbers = [1, 2, 3, 4, 5]
```

```python
new_list = [num + 1 for num in numbers]
new_list
```

```python
new_list2 = []
for num in numbers:
    new_list2.append(num + 1)

new_list2
```

```python
[num * 10 for num in numbers]
```

```python
[num for num in numbers if num > 1]
```

```python
[num * 10 for num in numbers if num > 1]
```

```python
[num for num in numbers if num % 2 != 0]
```

```python
for index, value in enumerate(numbers):
    print(index, value)
```

```python
[value for index, value in enumerate(numbers) if (index + 1) % 2 != 0]
```
